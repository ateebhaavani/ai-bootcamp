```python
import nltk
from nltk.tokenize import word_tokenize

nltk.download('punkt_tab')

text = """My hobbies are travelling and listening to music. I enjoy travelling because it gives me an opportunity to explore new places, experience different cultures, and learn about the world beyond my usual surroundings. Whenever I travel, I like visiting historical places, beautiful natural locations, and new cities. Travelling also helps me gain knowledge and develop a better understanding of different people and their lifestyles. I particularly enjoy places surrounded by mountains, beaches, greenery, and scenic views. It also gives me a break from my daily routine and helps me feel refreshed and motivated.

Another hobby I enjoy is listening to music. Music is an important part of my daily life because it helps me relax and improve my mood. I listen to different types of music depending on my mood and situation. I often listen to music while studying, travelling, or relaxing in my free time. Music also helps me concentrate and makes long or tiring activities more enjoyable.

Both travelling and listening to music help me maintain a balance between learning, relaxation, and entertainment. These hobbies allow me to discover new experiences, express myself, and enjoy my free time in a meaningful way. I believe hobbies are important because they refresh our minds and contribute to our overall personality and well-being.

"""

words = word_tokenize(text.lower())

print(words)
```

```python
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')

# Remove stopwords
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word.isalpha() and word not in stop_words]
# The general syntax for a list comprehension is: [expression for item in iterable if condition]
print(filtered_words)
```

```python
from collections import Counter

word_freq = Counter(filtered_words)
print(word_freq)
```

```python
from wordcloud import WordCloud

wordcloud = WordCloud(
    width=800,
    height=400,
    background_color='white'
).generate_from_frequencies(word_freq)
```

```python
import matplotlib.pyplot as plt

plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.savefig("wordcloud.jpg",format="jpg",dpi=300,bbox_inches="tight")
plt.show()
```
