```python
#program to create a list of random numbers and print all the even numbers.
numbers  = [1,2,4,5,8,7,89,34]
for i in range(0,len(numbers)-1):
    if numbers[i]%2==0:
        print("the even numbers from the list are:",numbers[i])
```

```python
#program to create a string and print it in reverse order
text = "python is a user friendly language"
#reverse the characters
print("reverse of string:",text[::-1])
#reverse the order of words
words = text.split()
print("reverse of words:"," ".join(words[::-1])) 
```
